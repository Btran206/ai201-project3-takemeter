```python
import pandas as pd
import re
import html
df = pd.read_csv("data/raw_posts.csv")
```

```python
df = df.dropna(subset=['body'])
```

```python
df = df.dropna(subset=['flair'])
```

```python
df = df.reset_index(drop = True)
```

```python
subsets = []
for i in ['hot', 'new', 'top', 'controversial']:
    subsets.append(df[df.source == i].iloc[0:50])
```

```python
df = pd.concat(subsets)
```

```python
df = df.reset_index(drop = True)
```

```python
df.to_csv("raw_posts.csv")
```

```python
def clean_text(text):
    # Decode HTML entities (e.g. &#x200B; &amp; &lt;)
    text = html.unescape(text)
    # Remove zero-width and other invisible Unicode spaces
    text = re.sub(r'[​‌‍­﻿]', '', text)
    # Fix escaped apostrophes/contractions like \'
    text = text.replace("\\'", "'")
    # Collapse multiple whitespace/newlines into a single space
    text = re.sub(r'\s+', ' ', text)
    return text.strip()

df['body'] = df['body'].apply(clean_text)
df['body'][0]
```

```python
df = df[["body", "label"]]
```

```python
df.to_csv("data/clean_posts.csv", index=False)
```
